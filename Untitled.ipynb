```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Read the image
img = cv2.imread("image.jpg")

# Check whether image is loaded
if img is None:
    print("Error: Image not found!")
else:
    print("Image loaded successfully")

# Convert BGR to RGB for displaying using Matplotlib
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Display original image
plt.figure(figsize=(6, 5))
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
# Averaging filter
average = cv2.blur(img, (5, 5))

# Convert BGR to RGB
average_rgb = cv2.cvtColor(average, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(6, 5))
plt.imshow(average_rgb)
plt.title("Averaging Filter")
plt.axis("off")
plt.show()
```

```python
# Weighted averaging kernel
kernel = np.array([
    [1, 2, 1],
    [2, 4, 2],
    [1, 2, 1]
], dtype=np.float32)

# Normalize the kernel
kernel = kernel / np.sum(kernel)

# Apply weighted averaging
weighted = cv2.filter2D(img, -1, kernel)

# Convert BGR to RGB
weighted_rgb = cv2.cvtColor(weighted, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(6, 5))
plt.imshow(weighted_rgb)
plt.title("Weighted Averaging Filter")
plt.axis("off")
plt.show()
```

```python
# Gaussian filter
gaussian = cv2.GaussianBlur(img, (5, 5), 0)

# Convert BGR to RGB
gaussian_rgb = cv2.cvtColor(gaussian, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(6, 5))
plt.imshow(gaussian_rgb)
plt.title("Gaussian Filter")
plt.axis("off")
plt.show()
```

```python
# Median filter
median = cv2.medianBlur(img, 5)

# Convert BGR to RGB
median_rgb = cv2.cvtColor(median, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(6, 5))
plt.imshow(median_rgb)
plt.title("Median Filter")
plt.axis("off")
plt.show()
```

```python
# Laplacian sharpening kernel
laplacian_kernel = np.array([
    [0, -1, 0],
    [-1, 5, -1],
    [0, -1, 0]
], dtype=np.float32)

# Apply sharpening kernel
laplacian_kernel_result = cv2.filter2D(img, -1, laplacian_kernel)

# Convert BGR to RGB
laplacian_kernel_rgb = cv2.cvtColor(
    laplacian_kernel_result, 
    cv2.COLOR_BGR2RGB
)

plt.figure(figsize=(6, 5))
plt.imshow(laplacian_kernel_rgb)
plt.title("Laplacian Sharpening Using Kernel")
plt.axis("off")
plt.show()
```

```python
# Convert image to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Apply Laplacian operator
laplacian = cv2.Laplacian(gray, cv2.CV_64F)

# Convert result to uint8
laplacian = cv2.convertScaleAbs(laplacian)

# Display
plt.figure(figsize=(6, 5))
plt.imshow(laplacian, cmap="gray")
plt.title("Laplacian Operator")
plt.axis("off")
plt.show()
```

```python
plt.figure(figsize=(15, 10))

# Original
plt.subplot(3, 3, 1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")

# Averaging
plt.subplot(3, 3, 2)
plt.imshow(average_rgb)
plt.title("Averaging")
plt.axis("off")

# Weighted Averaging
plt.subplot(3, 3, 3)
plt.imshow(weighted_rgb)
plt.title("Weighted Averaging")
plt.axis("off")

# Gaussian
plt.subplot(3, 3, 4)
plt.imshow(gaussian_rgb)
plt.title("Gaussian")
plt.axis("off")

# Median
plt.subplot(3, 3, 5)
plt.imshow(median_rgb)
plt.title("Median")
plt.axis("off")

# Laplacian Kernel
plt.subplot(3, 3, 6)
plt.imshow(laplacian_kernel_rgb)
plt.title("Laplacian Kernel Sharpening")
plt.axis("off")

# Laplacian Operator
plt.subplot(3, 3, 7)
plt.imshow(laplacian, cmap="gray")
plt.title("Laplacian Operator")
plt.axis("off")

plt.tight_layout()
plt.show()
```
